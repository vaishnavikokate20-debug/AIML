```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('train.txt.zip',sep=';',header=None,names=['text','emotions'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotions=df['emotions'].unique()
emotions_numbers={}
i=0
for emo in unique_emotions:
  emotions_numbers[emo]=i
  i +=1
df['emotions']=df['emotions'].map(emotions_numbers)
```

```python
df
```

```python
df['text']=df['text'].apply(lambda x:x.lower())
```

```python
import string
def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
def remove_numbers(txt):
  new=""
  for i in txt:
      if not i.isdigit():
        new=new+i
  return new
df['text']=df['text'].apply(remove_numbers)
```

```python
def remove_emojis(txt):
  new=""
  for i in txt:
      if i.isascii():
        new+=i
  return new
df['text']=df['text'].apply(remove_emojis)

```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words=set(stopwords.words('english'))
stop_words
```

```python
len(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
  words=txt.split()
  cleaned=[]
  for i in words:
    if not i in stop_words:
      cleaned.append(i)
  return ' '.join(cleaned)
```

```python
df['text']=df['text']. apply(remove)
```

```python
df.loc[1]['text']
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotions'], test_size=0.20, random_state=42)

```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

```

```python
bow_vectorizer=CountVectorizer()
```

```python
X_train_bow=bow_vectorizer.fit_transform(X_train)
X_test_bow=bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
nb_model=MultinomialNB()
nb_model.fit(X_train_bow,y_train)
```

```python
pred_bow=nb_model.predict(X_test_bow)
print(accuracy_score(y_test,pred_bow))
```

```python
pred_bow
```

```python
TfidfVectorizer=TfidfVectorizer()
```

```python
X_train_Tfidf=TfidfVectorizer.fit_transform(X_train)
X_test_Tfidf=TfidfVectorizer.transform(X_test)
```

```python
nb2_model=MultinomialNB()
nb2_model.fit(X_train_Tfidf,y_train)
```

```python
y_pred=nb2_model.predict(X_test_Tfidf)
```

```python
 print(accuracy_score(y_test,y_pred))
```

```python
log_pred=logistic_model.predict(X_test_Tfidf)
```
